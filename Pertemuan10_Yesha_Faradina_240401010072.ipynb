```python
#1
import pandas as pd
import kagglehub

path = kagglehub.dataset_download("blastchar/telco-customer-churn")

print(path)
df = pd.read_csv("/kaggle/input/telco-customer-churn/WA_Fn-UseC_-Telco-Customer-Churn.csv")
print(df.shape)
print(df["Churn"].value_counts(normalize=True))
```

```python
#2
from sklearn.model_selection import train_test_split

#TODO: encodeing fitur kategorikal (mis. pd.get_dummies)
#TODO: pisahkab X (fitur) dan y (target = Churn)

df = pd.get_dummies(df, drop_first=True)

X = df.drop("Churn_Yes", axis=1)
y = df ["Churn_Yes"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("Data latih :", X_tr.shape)
print("Data Uji :", X_te.shape)
```

```python
#3 - #4
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score

rf = RandomForestClassifier(n_estimators=300, class_weight="balanced", random_state=42)
rf.fit(X_tr, y_tr)
pred = rf.predict(X_te)
proba = rf.predict_proba(X_te)[:,1]
print(classification_report(y_te, pred))
print("ROC-AUC:", roc_auc_score(y_te, proba))
```

```python
#5

churn_prob = rf.predict_proba(X)[:, 1]

hasil = pd.DataFrame({
    "Churn_Probabilities": churn_prob,
    "Churn_Percentage(%)": (churn_prob * 100).round(2)
})

hasil.head(10)
```

Berdasarkan hasil prediksi Random Forest, setiap pelanggan memiliki probabilitas churn yang berbeda-beda. Beberapa pelanggan memiliki kemungkinan churn yang sangat tinggi, seperti 96,33%, 89,33%, dan 85,33%, sehingga perlu menjadi prioritas dalam program retensi pelanggan. Sebaliknya, terdapat pelanggan dengan probabilitas churn yang sangat rendah, seperti 0,67%, 2,67%, dan 1,33%, yang menunjukkan mereka cenderung tetap menggunakan layanan. Hasil prediksi probabilitas ini dapat membantu perusahaan mengidentifikasi pelanggan yang berisiko tinggi dan mengambil tindakan pencegahan untuk mengurangi tingkat churn.
